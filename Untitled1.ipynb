```python
import pandas as pd

data = {
    "Name": ["Chandra", "Ravi", "Arun", "Kiran", "Sai",
             "Rahul", "Vijay", "Akhil", "Suresh", "Priya"],

    "Age": [20, 21, 19, 22, 20, 21, 20, 19, 22, 20],

    "Department": ["CSE", "ECE", "CSE", "ECE", "CSE",
                   "EEE", "CSE", "ECE", "EEE", "CSE"],

    "Marks": [85, 72, 90, 65, 95, 78, 88, 70, 82, 92],

    "Attendance": [90, 75, 95, 68, 98, 80, 92, 72, 85, 96]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created!")
df.shape
df.info()
df[df["Attendance"] > 80].max()
df.sort_values(by="Attendance").head(3)
len(df[df["Attendance"] > 80]) * 100 / len(df)
len(df[(df["Attendance"] >= 70) & (df["Attendance"] <= 80)])
len(df[(df["Department"] == "CSE") &
       (df["Attendance"] >= 70) &
       (df["Attendance"] <= 80)]) * 100 / len(df[df["Department"] == "CSE"])
```
